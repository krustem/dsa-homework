## [База]
1. Напишите функции `to_num(s)` и `clean_city(s)` (6 городов, латиница и разный регистр → кириллица с заглавной).
2. Сколько заказов в каждой категории (`category`)? Отсортируйте по убыванию.
3. Для каждой категории посчитайте выручку оплаченных заказов и средний чек. Выведите таблицу f-строкой.
4. Сколько заказов в каждом месяце? Месяц — это `order_date[:7]`.

````python
# База
print(rows[:7])
#1
def to_num(s):
  return float(s.replace(" ", ""))

def clean_city(s):
  s = s.strip()
  s.lower().replace(" ", "").title()
  if s == "Almaty":
    s = "Алматы"
  if s == "Astana":
    s = "Астана"
  if s == "Shymkent":
    s = "Шымкент"
  if s == "Karaganda":
    s = "Караганда"
  return s


````

````python
print("Для первых 6 городов сумма выручок с использование очистки to_num:")
for i in rows[:7]:
  print(to_num(i['amount']))
````

````python
for i in rows[:7]:
  print(clean_city(i['city']))
````

````python

#2
num_cat = {}
count = 0
for i in rows[:7]:
  if num_cat.get(i['category'], 0) and i['status'] == 'оплачен':
    num_cat[i['category']] += 1
  else:
    if i['status'] == 'оплачен':
      num_cat[i['category']] = 1
print(num_cat)
print("Отсортированный:")
print(sorted(num_cat, key=num_cat.get, reverse=True))


#3
num_cat_sum = {}
count = 0
for i in rows[:7]:
  if num_cat_sum.get(i['category'], 0) and i['status'] == 'оплачен':
    num_cat_sum[i['category']] += to_num(i['amount'])
  else:
    if i['status'] == 'оплачен':
      num_cat_sum[i['category']] = to_num(i['amount'])
print(num_cat_sum)


num_cat_avg = {}
for i in num_cat_sum:
  num_cat_avg[i] = num_cat_sum[i]/num_cat[i]
  print(f"Средний чек по категории {i} составляет {num_cat_avg[i]}")
print("Средний чек общая картина:")
print(num_cat_avg)


#4
dict_month = {}
for i in rows:
  if dict_month.get(i['order_date'][:7], 0):
    dict_month[i['order_date'][:7]] += 1
  else:
    dict_month[i['order_date'][:7]] = 1
print("Сколько заказов в каком месяце:")
print(dict_month)
````
