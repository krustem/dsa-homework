````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# База
````

````python
#1
clean_cities = []
for city in cities:
  city = city.strip()
  clean_cities.append(city.upper())
clean_cities
````

````python
#2
amounts_int = []
for amount in amounts:
  amounts_int.append(int(amount))
amounts_int
````

````python
#3
print(f"количество заказов: {len(amounts)}")
print(f"общая сумма заказов: {sum(amounts_int)}")
print(f"количество заказов: {round(sum(amounts_int)/len(amounts_int))}")
````

````python
#4
not_paid = 0
not_paid_sum = 0
for i in range(len(paid)):
  if paid[i] == False:
    not_paid += 1
    not_paid_sum += amounts_int[i]

print(f"не оплачено заказов: {not_paid}")
print(f"сумма не оплаченных заказов: {not_paid_sum}")
````

````python
#5
for city, amount, status in zip(clean_cities, amounts_int, paid):
  print(f"{city} | {amount} | {status}")
````

````python
#7
def max_index(arr):
  krupniy = max(amounts_int)
  krup_index = amounts_int.index(krupniy)
  if paid[krup_index] == True:
    print(f"номер (индекс) самого крупного оплаченного заказа: {krup_index}")
    print(f"город самого крупного оплаченного заказа: {clean_cities[krup_index]}")
  else:
    del amounts_int[krup_index]
    del clean_cities[krup_index]
    del paid[krup_index]
    max_index(arr)

maxarr = amounts_int.copy()
max_index(maxarr)
````
